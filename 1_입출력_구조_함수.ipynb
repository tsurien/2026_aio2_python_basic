```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "코딩", "데이터",  "AI"]

found_word = None

for word in words: # 워드 안에 4개 있어서 4번 반복.
    if word in sentence: # 워드 안에 센텐스가 있으면
        found_word = word
        break

if found_word:
    print(f"입력한 문장에서 '{found_word}' 단어를 찾았습니다!")
else:
    print("입력한 문장에 해당 단어가 없습니다.")
```

# 함수

```python
def add_int(a, b):  # 디파인 : 함수 정의할땐 꼭 써야한다
    return a+b  # 파이썬은 다른거는 생략해주지만 함수는 써야한다
print(add_int(2,3))
```

```python
add_int() # 함수 호출
# 에러 나는 이유 : a,b가 없어서 다른 함수라고 인식함
# 시그니쳐
```
