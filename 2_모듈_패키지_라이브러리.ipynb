```python
import pandas as pd
# 표로 보기위한 원시 데이터(딕셔너리(사전))
data = {
    '이름': ['김철수', '이영희', '박민수', '정수진', '최영호', '한미래', '윤지원'],
    '나이': [25, 28, 22, 31, 27, 24, 29],
    '점수': [85, 92, 78, 95, 88, 73, 91]
}
# 판다스의 데이터프레임객체 생성해서 표로 보기좋게 만들어줌
df = pd.DataFrame(data)
df # 상위 5개만 꺼내주는 메소드
# 인덱스는 0부터 시작!
int(df["나이"].mean())
```
